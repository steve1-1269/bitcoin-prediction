```python
import pandas as pd
import matplotlib.pyplot as plt
import joblib
```

```python
model= joblib.load(r"C:\Users\MINE\Desktop\bitcoin prediction\expo_smoo_forecast.pkl")
```

```python
df= pd.read_csv(r"C:\Users\MINE\Desktop\bitcoin prediction\expo_smoo_forecast.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
df.set_index('Date', inplace=True)
```

```python
df.info()
```

```python
df.index= pd.to_datetime(df.index)
```

```python
# Predicting the next 50 days price 

predictions= model.forecast(50)
```

```python
predictions
```

```python
plt.figure(figsize=(10,4))
plt.plot(df.Y, label= 'Train')
plt.plot(predictions, label= 'predictions')

plt.tight_layout()
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.plot(df.Y.loc['2026'], label= 'Train')
plt.plot(predictions, label= 'predictions')

plt.tight_layout()
plt.legend()
plt.show()
```
